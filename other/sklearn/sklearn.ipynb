### **Regressão Polinomial**

PolynomialFeatures(degree=k) é um transformer que recebe ate que grau queremos computar as features polinomiais, ou seja, se degree é 2 criamos features que combinadas com as existentes de ate grau 2 como por exemplo x1^2, x2^2 e x1x2. Para usar o objeto o .fit(X) nos dados de treino tem que ser feito e em seguida o .transform(X) para criar os dados de treino novos. O objeto do treino tem que ser reutilizado na inferencia com o .transform(X) pois é um pre-processamento e precisa manter o padrão que todos os outros tem para evitar data leak mesmo que não exista data leak na criação de features polinomiais além disso, se no teste tiver numero de features diferentes a quantidade de features criada vai ser diferente se fizer o .fit(X) no teste.

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

X_train = []
y_train = []
X_test = []

poly_features = PolynomialFeatures(degree=4)
poly_features.fit(X=X_train)
poly_X_train = poly_features.transform(X=X_train)

poly_reg = LinearRegression()
poly_reg.fit(poly_X_train, y_train)

poly_X_test = poly_features.transform(X_test)
y_test = poly_reg.predict(poly_X_test)
```
