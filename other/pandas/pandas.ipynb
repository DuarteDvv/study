
```sql
SELECT
  e.app_id,
  ROUND(
    100.0 * SUM(CASE WHEN e.event_type = 'click' THEN 1 ELSE 0 END) 
    / 
    SUM(CASE WHEN e.event_type = 'impression' THEN 1 ELSE 0 END)
  , 2) AS ctr
FROM
  events AS e
WHERE
  e.timestamp >= '2022/01/01' AND e.timestamp < '2023/01/01'
GROUP BY  
  e.app_id;
```
